```python
print("Missing values in each column:")
print(raw.isna().sum())
```

```python
print("Exact duplicates in original dataset:", raw.duplicated().sum())
```

```python
clean = raw.copy()

print("Working copy created.")
print("Rows in working copy:", clean.shape[0])
```

```python
clean = clean.drop_duplicates()

print("Rows after removing exact duplicates:", clean.shape[0])
print("Exact duplicates remaining:", clean.duplicated().sum())
```

```python
clean["item"] = clean["item"].str.strip().str.title()

print("Item names after cleaning:")
print(clean["item"].value_counts())
```

```python
print("City names:")
print(clean["city"].value_counts())
```

```python
print("Units:")
print(clean["unit"].value_counts())
```

```python
clean["price"] = clean["price"].str.replace(",", "", regex=False)

print("Price values after removing commas:")
print(clean["price"].head(10))
```

```python
clean["price"] = pd.to_numeric(clean["price"], errors="coerce")

print("Price data type:", clean["price"].dtype)
print("Missing prices after conversion:", clean["price"].isna().sum())
```

```python
display(clean.loc[17, ["date", "item", "city", "price"]])
```

```python
clean["date"] = pd.to_datetime(clean["date"])

print("Date data type:", clean["date"].dtype)
```

```python
print(clean["date"].head())
```

```python
repeated_keys = clean[clean.duplicated(
    subset=["date", "item", "city"],
    keep=False
)]

print("Rows with repeated date-item-city combinations:")
display(repeated_keys)
```

```python
print(
    "Number of repeated date-item-city combinations:",
    clean.duplicated(subset=["date", "item", "city"]).sum()
)
```

```python
print("Exact duplicates remaining:", clean.duplicated().sum())
```

```python
print("Missing values after cleaning:")
print(clean.isna().sum())
```

```python
print("Final column data types:")
print(clean.dtypes)
```

```python
print("Final item counts:")
print(clean["item"].value_counts())
```

```python
print("Original dataset shape:", raw.shape)
print("Cleaned dataset shape:", clean.shape)
```

```python
print("Missing values in original dataset:")
print(raw.isna().sum())

print("\nMissing values in cleaned dataset:")
print(clean.isna().sum())
```

```python
summary = pd.DataFrame({
    "Problem": [
        "Exact duplicate row",
        "Inconsistent item capitalization",
        "Price containing a comma",
        "Missing price",
        "Dates stored as text",
        "Repeated date-item-city combination"
    ],
    "Action Taken": [
        "Removed exact duplicate",
        "Standardized item names",
        "Removed comma and converted price to numeric",
        "Kept missing value as NaN",
        "Converted dates to datetime",
        "Investigated records and retained both"
    ],
    "Result": [
        "One duplicate removed",
        "Item names standardized",
        "Price 1,250 converted to 1250.0",
        "One missing price remains",
        "Dates converted successfully",
        "Two different prices found for the same key"
    ]
})

display(summary)
```

```python
print("FINAL DATASET CHECK")
print("-------------------")

print("Original rows:", raw.shape[0])
print("Cleaned rows:", clean.shape[0])

print("Original columns:", raw.shape[1])
print("Cleaned columns:", clean.shape[1])

print("Exact duplicates remaining:", clean.duplicated().sum())

print("Missing values:")
print(clean.isna().sum())

print("\nData types:")
print(clean.dtypes)
```

### Before Cleaning
The dataset contained 97 rows and 6 columns, with one exact duplicate, one missing price, inconsistent item capitalization, a price containing a comma, and dates stored as text.

### After Cleaning
The dataset contains 96 rows and 6 columns. Exact duplicates were removed, item names standardized, prices and dates converted to appropriate data types, and the missing price and conflicting records retained for investigation.
